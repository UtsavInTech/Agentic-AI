***OPENAI***

```python
from openai import OpenAI
client = OpenAI()

response = client.responses.create(
    model="gpt-5-nano",
    input="Write a one-sentence bedtime story about a unicorn."
)

print(response.output_text)
```

***Gemini***

```python
from google import genai

client = genai.Client()

response = client.models.generate_content(
    model="gemini-3-flash-preview",
    contents="Explain how AI works in a few words",
)

print(response.text)
```

***Anthropic***

```python
import os
from anthropic import Anthropic

client = Anthropic(
    api_key=os.environ.get("ANTHROPIC_API_KEY"),  # This is the default and can be omitted
)
message = client.messages.create(
    max_tokens=1024,
    messages=[{
        "content": "Hello, world",
        "role": "user",
    }],
    model="claude-sonnet-4-5-20250929",
)
print(message.id)
```

***LangChain***

```python
from langchain.chat_openai import ChatOpenAI

model = ChatOpenAI(model_name="gpt-4", temperature=0)
response = model.invoke("Tell me a joke about computers.")
print(response)
```

```python
from langchain.chat_anthropic import ChatAnthropic

model = ChatAnthropic(model_name="claude-sonnet-4-5-20250929", temperature=0)
response = model.invoke("Tell me a joke about computers.")
print(response)
```

```python
from langchain.chat_gemini import ChatGemini

model = ChatGemini(model_name="gemini-3-flash-preview", temperature=0)
response = model.invoke("Tell me a joke about computers.")
print(response)
```
